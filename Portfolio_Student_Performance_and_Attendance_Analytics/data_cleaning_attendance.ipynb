```python
import pandas as pd
```

```python
# Inisialisasi Data
df = pd.read_csv('data_raw/attendance.csv', sep=',')
df.columns = df.columns.str.strip()
```

```python
# Mengubah Format
# Mengubah Format Date (memastikan format dengan benar)
df['Date'] = pd.to_datetime(df['Date'], format='mixed', dayfirst=True)

# Mengubah Format Subject dan Attendance_Status (memastikan setiap awal kata huruf kapital)
df['Subject'] = df['Subject'].str.title()
df['Attendance_Status'] = df['Attendance_Status'].str.title()

df.head(15)
```

```python
# Mengecek Duplikat dan Info
print(f'Total baris duplikat: {df.duplicated().sum()}')
df.drop_duplicates(inplace=True)
print(f'Total baris setelah menghapus duplikat: {df.duplicated().sum()}')

df.info()

df.to_csv('data_clean/clean_attendance_data.csv', index=False)
```
