```python
import pandas as pd
```

```python
# Inisialisasi Data
df = pd.read_csv('data_raw/teacher_parent_communication.csv', sep=',')
df.columns = df.columns.str.strip()
```

```python
# Data Transformation
# Memastikan Format pada Kolom Date Sesuai
df['Date'] = pd.to_datetime(df['Date'], format='mixed', dayfirst=True)

# Mengisi Nilai Kosong pada Kolom Message_Content
df['Message_Content'] = df['Message_Content'].str.strip()
df['Message_Content'] = df['Message_Content'].replace('', '-')
df['Message_Content'] = df['Message_Content'].fillna('-')

df.head(15)
```

```python
# Mengecek Duplikat, Jumlah Data pada Dataset, dan Info
print(f'Total baris duplikat: {df.duplicated().sum()}')
df.drop_duplicates(inplace=True)
print(f'Total baris setelah menghapus duplikat: {df.duplicated().sum()}')

print(f'Total baris data yang kosong: {df.isna().sum().sum()}')
print(f'Total data pada dataset Attendance: {df.shape} [baris, kolom]')

df.info()

df.to_csv('data_clean/clean_teacher_parent_communication_data.csv', index=False)
```
