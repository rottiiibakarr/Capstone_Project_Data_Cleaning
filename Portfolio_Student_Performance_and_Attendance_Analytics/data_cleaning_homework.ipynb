```python
import pandas as pd 
import numpy as np
```

```python
# Inisialisasi Data
df = pd.read_csv('data_raw/homework.csv', sep=',')
df.columns = df.columns.str.strip()
```

```python
# Mengubah Format
# Mengubah Format Subject dan Assignment_Name (memastikan format dengan benar)
df['Subject'] = df['Subject'].str.title()
df['Assignment_Name'] = df['Assignment_Name'].str.title()

# Mengubah Format Due_Date
df['Due_Date'] = pd.to_datetime(df['Due_Date'], format='mixed', dayfirst=True)

df.head(15)
```

```python
# Feature Engineering
# Mengubah Data yang Tidak Sesuai pada Kolom Status
df['Status'] = df['Status'].str.title()

# Mencegah error jika ada baris yang kosong (None/NaN) dengan contains -> na=False
condition_status = [
    df['Status'].str.contains('✅|✔', na=False),
    df['Status'].str.contains('❌', na=False)
]

choice_status = ['Done', 'Not Done']

# Menerapkan perubahan, defaultnya mempertahankan nilai asli jika tidak ada simbol
df['Status'] = np.select(condition_status, choice_status, default=df['Status'])

# Mengubah Nilai yang Kosong (None/NaN) menjadi - 
df['Guardian_Signature'] = df['Guardian_Signature'].str.strip()
df['Guardian_Signature'] = df['Guardian_Signature'].replace('', '-')

df.head(15)
```

```python
# Mengecek Duplikat dan Info
print(f'Total baris duplikat: {df.duplicated().sum()}')
df.drop_duplicates(inplace=True)
print(f'Total baris setelah menghapus duplikat: {df.duplicated().sum()}')

df.info()

df.to_csv('data_clean/clean_homework_data.csv', index=False)
```
