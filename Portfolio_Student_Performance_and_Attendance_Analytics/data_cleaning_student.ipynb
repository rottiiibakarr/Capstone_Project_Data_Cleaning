```python
import pandas as pd
import numpy as np
```

```python
# Inisialisasi Data
df = pd.read_csv('data_raw/students.csv', sep=',')
df.columns = df.columns.str.strip()
```

```python
# Data Transformation
# Mengubah Format Nama dengan Kapital setiap Kata
df['Full_Name'] = df['Full_Name'].str.title()

# Mengubah Format Tanggal pada Kolom Date_of_Birth
df['Date_of_Birth'] = pd.to_datetime(df['Date_of_Birth'], format='mixed', dayfirst=True)

# Mengubah Format pada Kolom Emergency_Contact dan Data Cleaning
df['Emergency_Contact'] = df['Emergency_Contact'].str.strip().str.replace('+1', '')
df['Emergency_Contact'] = df['Emergency_Contact'].str.replace('(', '').str.replace(')', '').str.replace('-', '').str.replace('.', '')
df['Emergency_Contact'] = df['Emergency_Contact'].str.replace(r'x.*', '', regex=True)
df['Emergency_Contact'] = '+1' + df['Emergency_Contact'].astype(str)
df['Emergency_Contact'] = df['Emergency_Contact'].str.replace('+1nan', '-')

df.head(15)
```
