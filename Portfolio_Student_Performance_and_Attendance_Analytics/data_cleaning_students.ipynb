```python
import pandas as pd
import numpy as np
```

```python
# Inisialisasi Data
df = pd.read_csv('data_raw/students.csv', sep=',')
df.columns = df.columns.str.strip()
```

```python
# Feature Engineering
# Membuat Kolom Baru dengan Nama Emergency_Contact_Ext 
df['Emergency_Contact_Ext'] = df['Emergency_Contact'].str.split('x').str[1].fillna('-')

df.head(15)
```

```python
# Data Transformation
# Mengubah Format Nama dengan Kapital setiap Kata
df['Full_Name'] = df['Full_Name'].str.title()

# Mengubah Format Tanggal pada Kolom Date_of_Birth
df['Date_of_Birth'] = pd.to_datetime(df['Date_of_Birth'], format='mixed', dayfirst=True)

# Mengubah Format pada Kolom Emergency_Contact dan Data Cleaning
df['Emergency_Contact'] = df['Emergency_Contact'].str.split('x').str[0]
df['Emergency_Contact'] = df['Emergency_Contact'].str.replace(r'^(\+1|001)', '', regex=True) # Hapus awalan '+1' ATAU '001' jika ada di bagian paling depan string
df['Emergency_Contact'] = df['Emergency_Contact'].str.replace(r'\D', '', regex=True) # Menghapus semua angka yang bukan angka (titik, strip, kurung, spasi)
df['Emergency_Contact'] = '+1' + df['Emergency_Contact'] 
df['Emergency_Contact'] = df['Emergency_Contact'].fillna('-')

df.head(15)
```

```python
# Mengecek Duplikat, Jumlah Data pada Dataset, dan Info
print(f'Total baris duplikat: {df.duplicated().sum()}')
df.drop_duplicates(inplace=True)
print(f'Total baris setelah menghapus duplikat: {df.duplicated().sum()}')

print(f'Total data pada dataset Attendance: {df.shape} [baris, kolom]')

df.info()

df.to_csv('data_clean/clean_students_data.csv', index=False)
```
