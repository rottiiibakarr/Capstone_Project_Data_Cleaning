```python
import pandas as pd
```

```python
# Inisialisasi Data
df = pd.read_csv('data_raw/performance.csv', sep=',')
df.columns = df.columns.str.strip()
```

```python
# Data Transformation
# Mengubah Format Subject (memastikan format dengan benar)
df['Subject'] = df['Subject'].str.strip().str.title()

# Mengubah Format Nilai Maksimal dan Minimal pada Kolom Exam_Score
df['Exam_Score'] = df['Exam_Score'].clip(lower=0, upper=100)

# Mengubah Format Tipe Data pada Kolom Homework_Completion_%
df['Homework_Completion_%'] = df['Homework_Completion_%'].astype(str).str.replace('%', '').astype(int)
df['Homework_Completion_%'] = df['Homework_Completion_%'].clip(lower=0, upper=100)

# Menangani Nilai yang Tidak Terinput (Missing Values)
df['Teacher_Comments'] = df['Teacher_Comments'].str.strip()
df['Teacher_Comments'] = df['Teacher_Comments'].replace('', '-')

df.head(30)
```

```python
# Mengecek Duplikat, Jumlah Data pada Dataset, dan Info
print(f'Total baris duplikat: {df.duplicated().sum()}')
df.drop_duplicates(inplace=True)
print(f'Total baris setelah menghapus duplikat: {df.duplicated().sum()}')

print(f'Total data pada dataset Attendance: {df.shape} [baris, kolom]')

df.info()

df.to_csv('data_clean/clean_performance_data.csv', index=False)
```
