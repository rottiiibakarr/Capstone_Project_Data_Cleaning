```python
import pandas as pd
```

```python
# Inisialisasi Data
df = pd.read_csv('data_raw/performance.csv', sep=',')
df.columns = df.columns.str.strip()
```

```python
# Mengubah Format
# Mengubah Format Subject dengan Memastikan Format Benar
df['Subject'] = df['Subject'].str.strip()

# Mengubah Nilai Maksimal dan Minimal pada Kolom Exam_Score
df['Exam_Score'] = df['Exam_Score'].clip(lower=0, upper=100)

df.head(20)
```
