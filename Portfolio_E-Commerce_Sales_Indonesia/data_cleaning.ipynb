```python
import pandas as pd
```

```python
# Inisialisasi Data
df = pd.read_csv('data_raw/indonesia_ecommerce_sales_challenging.csv', sep=';')
df.columns = df.columns.str.strip()
```

```python
# Pengumpulan dan Pemahaman Data
df.info()
display(df.describe())

print(f'Total data pada dataset: {df.shape} [baris, kolom]')
```

```python
# Menghapus Duplikasi
print(f'Total baris duplikat: {df.duplicated().sum()}')
df.drop_duplicates(subset=['order_id'])

print(f'Total baris setelah menghapus duplikat: {df.duplicated().sum()}')

df.head(10)
```

```python
# Menangani Nilai Hilang
print(f'Jumlah data yang hilang pada dataset:\n{df.isnull().sum()} \n')

df = df.dropna(subset=['order_date', 'total_qty'])

df['Metode Pembayaran'] = df['Metode Pembayaran'].fillna('Unknown')

print(f'Jumlah data yang hilang pada dataset setelah ditangani:\n{df.isnull().sum()} \n')

df.head(10)
```

```python
# Standarisasi
df = df.rename(columns={
    'total_weight_gr': 'total_berat_gr',
    'total_returned_qty': 'total_pengembalian_qty',
    'product_categories': 'kategori_produk',
    'num_product_categories': 'jumlah_kategori_produk',
    'Total Diskon': 'total_diskon',
    'Status Pesanan': 'status_pesanan',
    'Alasan Pembatalan': 'alasan_pembatalan',
    'Opsi Pengiriman': 'opsi_pengiriman',
    'Metode Pembayaran': 'metode_pembayaran',
    'Pengirim Dibayar oleh Pembeli': 'pengirim_dibayar_pembeli',
    
    'order_date': 'tanggal_pesanan'
})

df = df.astype({
    'total_qty': 'int',
    'total_pengembalian_qty': 'int',
})

df['Waktu Pesanan Dibuat'] = pd.to_datetime(df['Waktu Pesanan Dibuat'], format='%Y-%m-%d %H:%M', errors='coerce')
df['tanggal_pesanan'] = pd.to_datetime(df['tanggal_pesanan'], format='mixed', dayfirst=True, errors='coerce')

print(f'Kategori produk pada dataset:\n{df['kategori_produk'].unique()}')
print(f'Jenis status pesanan pada dataset:\n{df['status_pesanan'].unique()}')
print(f'Alasan pembatalan pada dataset:\n{df['alasan_pembatalan'].unique()}\n')

df['alasan_pembatalan'] = df['alasan_pembatalan'].replace({
    'Dibatalkan oleh Pembeli. Alasan: Ubah Pesanan yang Ada': 'Dibatalkan oleh Pembeli',
    'Dibatalkan oleh Pembeli. Alasan: Lainnya/ berubah pikiran': 'Dibatalkan oleh Pembeli',
    'Dibatalkan secara otomatis oleh sistem. Alasan: Pengiriman gagal': 'Dibatalkan secara otomatis oleh sistem',
    'Dibatalkan secara otomatis oleh sistem. Alasan: Penjual tidak mengatur pengiriman tepat waktu': 'Dibatalkan secara otomatis oleh sistem',
    'Dibatalkan oleh Pembeli. Alasan: Need to change delivery address': 'Dibatalkan oleh Pembeli',
    'Dibatalkan secara otomatis oleh sistem. Alasan: Pesanan belum dibayar': 'Dibatalkan secara otomatis oleh sistem',
    'Dibatalkan secara otomatis oleh sistem. Alasan: Penjual gagal mengirimkan pesanan tepat waktu': 'Dibatalkan secara otomatis oleh sistem', 
    'Dibatalkan oleh Pembeli. Alasan: Perlu mengubah alamat pengiriman': 'Dibatalkan oleh Pembeli',
    'Dibatalkan oleh Pembeli. Alasan: Perlu mengubah pesanan': 'Dibatalkan oleh Pembeli',
    'Dibatalkan oleh Penjual. Alasan: Produk habis': 'Dibatalkan oleh Penjual',
    'Dibatalkan oleh Pembeli. Alasan: Menemukan yang lebih murah':'Dibatalkan oleh Pembeli',
    'Dibatalkan oleh Pembeli. Alasan: Penjual tidak responsif terhadap pertanyaan Pembeli': 'Dibatalkan oleh Pembeli',
    'Dibatalkan oleh Pembeli. Alasan: Proses pembayaran sulit': 'Dibatalkan oleh Pembeli',
    'Dibatalkan oleh Pembeli. Alasan: Lainnya': 'Dibatalkan oleh Pembeli',
    'Dibatalkan secara otomatis oleh sistem. Alasan: Paket hilang di perjalanan. Kompensasi yang memenuhi syarat telah dikreditkan ke Saldo Penjual-mu.': 'Dibatalkan secara otomatis oleh sistem',
    'Dibatalkan oleh Pembeli. Alasan: Perlu mengubah Voucher': 'Dibatalkan oleh Pembeli', 
    'Dibatalkan oleh Pembeli. Alasan: Tidak ingin membeli lagi': 'Dibatalkan oleh Pembeli',
    'Dibatalkan secara otomatis oleh sistem. Alasan: Lainnya': 'Dibatalkan secara otomatis oleh sistem'
})

print(f'Alasan pembatalan setelah diperbaiki pada dataset:\n{df['alasan_pembatalan'].unique()}')

df.head(10)
```
