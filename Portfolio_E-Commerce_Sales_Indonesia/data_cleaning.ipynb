```python
import pandas as pd
```

```python
# Inisialisasi Data
df = pd.read_csv('data_raw/indonesia_ecommerce_sales_challenging.csv', sep=';')
df.columns = df.columns.str.strip()
```

```python
# Pengumpulan dan Pemahaman Data
df.info()
display(df.describe())

print(f'Total data pada dataset: {df.shape} [baris, kolom]')
```

```python
# Menghapus Duplikasi
print(f'Total baris duplikat: {df.duplicated().sum()}')
df.drop_duplicates(subset=['order_id'])

print(f'Total baris setelah menghapus duplikat: {df.duplicated().sum()}')

df.head(10)
```

```python
# Menangani Nilai Hilang
print(f'Jumlah data yang hilang pada dataset:\n{df.isnull().sum()} \n')

df = df.dropna(subset=['order_date', 'total_qty'])

df['Metode Pembayaran'] = df['Metode Pembayaran'].fillna('Unknown')

print(f'Jumlah data yang hilang pada dataset setelah ditangani:\n{df.isnull().sum()} \n')

df.head(10)
```

```python
# Standarisasi
df = df.rename(columns={
    'total_weight_gr': 'total_berat_gr',
    'total_returned_qty': 'total_pengembalian_qty',
    'product_categories': 'kategori_produk',
    'num_product_categories': 'jumlah_kategori_produk',
    'order_date': 'tanggal_pesanan'
})

df.head(10)
```
