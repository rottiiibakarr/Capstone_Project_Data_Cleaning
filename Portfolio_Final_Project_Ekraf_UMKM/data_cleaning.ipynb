```python
import pandas as pd
```

```python
# Inisialisasi Data
df = pd.read_csv('data_raw/synthetic_umkm_data.csv', sep=',')
df.columns = df.columns.str.strip()
```

```python
# Inspeksi & Profiling Data
df.info()
display(df.describe())

print(f'Total data pada dataset: {df.shape} [baris, kolom]')
```

```python
# Menghapus Duplikasi (Removing Duplicates)
print(f'Total baris duplikat: {df.duplicated().sum()}')
df.drop_duplicates()

print(f'Total baris setelah menghapus duplikat: {df.duplicated().sum()}')

df.head(10)
```

```python
# Menangani Nilai Hilang (Missing Value)
print(f'Jumlah data yang hilang pada dataset:\n{df.isnull().sum()} \n')

df.head(10)
```

```python
# Standarisasi (Standardization)
print(f'Kombinasi latensi jam sibuk pada dataset UMKM: {df['Peak_Hour_Latency'].unique()}')
print(f'Kombinasi jenis kelas pada dataset UMKM: {df['Class'].unique()}')



df.head()
```
