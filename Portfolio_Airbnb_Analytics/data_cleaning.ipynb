```python
import pandas as pd
```

```python
df = pd.read_csv('data_raw/Airbnb_Cleaned_Europe_Dataset.csv', sep=',')
df.columns = df.columns.str.strip() # Menghapus whitespaces pada kolom/header tabel (jika ada)

# Mengecek jumlah baris apakah terdapat duplikat atau tidak sebelum proses cleaning
print(f'Total baris duplikat: {df.duplicated().sum()}')
```

```python
# Mengubah Format 
# Mengubah Format Jumlah Desimal 
df['Price'] = df['Price'].round(2)
df['City Center (km)'] = df['City Center (km)'].round(2)
df['Metro Distance (km)'] = df['Metro Distance (km)'].round(2)
df['Attraction Index'] = df['Attraction Index'].round(2)
df['Normalised Attraction Index'] = df['Normalised Attraction Index'].round(2)
df['Restraunt Index'] = df['Restraunt Index'].round(2)
df['Normalised Restraunt Index'] = df['Normalised Restraunt Index'].round(2)

# Mengubah Tipe Data 
df['Person Capacity'] = df['Person Capacity'].astype(int)
df['Cleanliness Rating'] = df['Cleanliness Rating'].astype(int)
df['Guest Satisfaction'] =df['Guest Satisfaction'].astype(int)

# Mengubah Setiap Kata menjadi Huruf Kapital
df['Room Type'] = df['Room Type'].str.title()

df.head()
```

```python
# Pengecekan Logika
# Pengecekan Konsistensi Tipe Kamar
df['Shared Room'] = df['Room Type'] == 'Shared Room'
df['Private Room'] = df['Room Type'] == 'Private Room'

# Pengecekan Batas Maksimal Rating
df['Cleanliness Rating'] = df['Cleanliness Rating'].clip(lower=0, upper=10)
df['Guest Satisfaction'] = df['Guest Satisfaction'].clip(lower=0, upper=100)

df.head()
```

```python
# Mengecek jumlah duplikat setelah dibulatkan
print(f'Duplikat setelah proses cleaning: {df.duplicated().sum()}')

# Menghapus baris yang terduplikasi
df.drop_duplicates(inplace=True)

# Pengecekan, apakah terdapat nilai yang masih tidak sesuai (contoh nilai minus)
df.describe()
```

```python
df.to_csv('data_clean/clean_airbnb_data.csv', index=False)
```
