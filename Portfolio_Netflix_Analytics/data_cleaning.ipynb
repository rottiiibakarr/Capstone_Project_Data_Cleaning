```python
import pandas as pd
```

```python
# Inisialisasi Data
df = pd.read_csv('data_raw/mymoviedb.csv', sep=',', lineterminator='\n')
df.columns = df.columns.str.strip()
```

```python
# Data Transformation
df['Original_Language'] = df['Original_Language'].str.upper()

df.head(10)
```

```python
# Mengecek Duplikat, Jumlah Data pada Dataset, dan Info
print(f'Total baris duplikat: {df.duplicated().sum()}')
df.drop_duplicates(inplace=True)
print(f'Total baris setelah menghapus duplikat: {df.duplicated().sum()}')

print(f'Total data pada dataset Attendance: {df.shape} [baris, kolom]')

df.info()

df.to_csv('data_clean/clean_mymoviedb_data.csv', index=False)
```
