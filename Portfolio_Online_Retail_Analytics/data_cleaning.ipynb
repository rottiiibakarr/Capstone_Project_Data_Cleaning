```python
import pandas as pd
import numpy as np
```

```python
# Inisialisasi Data
df = pd.read_csv('data_raw/online_retail_real_world.csv', sep=',')
df.columns = df.columns.str.strip()
```

```python
# Mengecek Duplikat, Jumlah Data pada Dataset, dan Info
print(f'Total baris duplikat: {df.duplicated().sum()}')
df.drop_duplicates(inplace=True)
print(f'Total baris setelah menghapus duplikat: {df.duplicated().sum()}')

print(f'Total data pada dataset Attendance: {df.shape} [baris, kolom]')

df.info()
```

```python
# Data Cleaning
df.fillna({
    'ProductName': '-',
    'Brand': '-',
    'Raw_Weight': '-',
    'Country': '-'
}, inplace=True)

# Untuk mendeteksi angka, simbol, tanda kurung/siku, dan sebagainya
regex_pattern_product = r'[!@#\$%\^&\*\(\)\[\]_\-\[\]\{\};:\'",.<>?/\\|`~+\d]'
regex_pattern_weight = r'[\!@#\$%\^&\*\(\)\[\]_\-\[\]\{\};:\'",.<>?/\\|`~+\D]'

df['ProductName'] = df['ProductName'].str.title().str.replace(regex_pattern_product, '', regex=True)
df['ProductName'] = df['ProductName'].str.replace(r'\s+', ' ', regex=True)

# Ekstraksi brand utama
df['Brand'] = df['Brand'].astype(str).str.split(',').str[0].str.strip() 
# Menstandarkan nilai yang awalnya kosong (NaN) yang ikut berubah menjadi string 'nan'
df['Brand'] = df['Brand'].replace('nan', '-') 
# Mengubah per-kata menjadi huruf kapital
df['Brand'] = df['Brand'].str.title()
# Mengganti tanda "’" menjadi "'" dan " and " menjadi " & "
df['Brand'] = df['Brand'].str.replace('’', "'").str.replace(' And ', ' & ')

# Mengganti satuan berat menjadi string kosong
df['Raw_Weight'] = df['Raw_Weight'].str.replace('gram', '').str.replace('g', '').str.replace('ml', '')
df['Raw_Weight'] = df['Raw_Weight'].str.strip().str.replace(regex_pattern_weight, '', regex=True)
df['Raw_Weight'] = pd.to_numeric(df['Raw_Weight'], errors='coerce')

# Menghitung nilai median dari kolom dataset Raw_Weight
median_raw_weight = df['Raw_Weight'].median()
# Menghitung median spesifik untuk masing-masing Brand
median_on_brand = df.groupby('Brand')['Raw_Weight'].transform('median')

df['Raw_Weight'] = np.where(
    df['Raw_Weight'].isna(), # Jika Raw_Weight kosong (NaN)
    np.where(
        df['Brand'] != '-', # Mengecek apakah kolom Brand diketahui (bukan NaN)
        median_on_brand,
        median_raw_weight
    ),
    df['Raw_Weight'] # Jika Raw_Weight tidak kosong, biarkan nilai aslinya
)

"""
Jika terdapat sebuah Brand baru yang semua datanya kebetulan kosong?
Maka median_on_brand untuk brand tersebut akan bernilai NaN juga.
Mengambil sisa NaN tersebut dan isi dengan median keseluruhan.
"""
df['Raw_Weight'] = df['Raw_Weight'].fillna(median_raw_weight)
# Tampilkan hasil pengecekan (seharusnya 0)
print("Jumlah data kosong setelah imputasi hybrid:", df['Raw_Weight'].isna().sum())
    
# Mengmbil kata pertama sebelum koma sebagai negara utama
df['Country'] = df['Country'].astype(str).str.split(',').str[0]
# r'^[a-z]{2}:' mencari dua huruf kecil di awal yang diikuti titik dua
df['Country'] = df['Country'].str.replace(r'^[a-z]{2}:', '', regex=True)
df['Country'] = df['Country'].str.strip().str.title()
# Standarisasi untuk negara yang bahasanya berantakan
df['Country'] = df['Country'].replace({
    'Argelia': 'Algeria',
    'Algérie': 'Algeria',
    'Austriche': 'Austria',
    'Belgien': 'Belgia',
    'Bélgica': 'Belgium',
    'Belgique': 'Belgium',
    'Bulgarie': 'Bulgaria',
    'España': 'Spain',
    'Francia': 'France',
    'Frankreich': 'France',
    'Ma': 'Morocco',
    'Maroc': 'Morocco',
    'Royaume Uni': 'United Kingdom'
})
df['Country'] = df['Country'].str.replace('-', ' ')

df['OrderDate'] = pd.to_datetime(df['OrderDate'], format='mixed', dayfirst=True)

df['UnitPrice'] = pd.to_numeric(df['UnitPrice'], errors='coerce')
df.head(20)
```

```python
# Konversi ke CSV
df.to_csv('data_clean/clean_online_retail_real_world_data.csv', index=False)
```
